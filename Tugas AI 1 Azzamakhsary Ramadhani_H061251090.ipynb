# 5 Prompt Geofisika dengan 8 Unsur

Dalam pembuatan prompt untuk pekerjaan geofisika, terdapat 8 unsur yang dapat digunakan agar instruksi lebih jelas dan terarah, yaitu **Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Optional Verification**.

## 1. Role (R)

**Role** adalah peran atau sudut pandang yang diberikan kepada AI dalam mengerjakan tugas. Dalam geofisika, role dapat berupa ahli atau peneliti pada bidang tertentu.

**Contoh:** geophysicist, seismologist, geophysical data analyst, atau reviewer metode geofisika.

## 2. Task (T)

**Task** adalah tugas utama yang harus dilakukan AI terhadap suatu data atau permasalahan geofisika.

**Contoh:** menganalisis data seismik, mengolah data gravitasi, menghitung anomali Bouguer, atau menginterpretasikan data magnetik.

## 3. Context (C)

**Context** menjelaskan latar belakang dan tujuan dari tugas yang dilakukan.

**Contoh:** data digunakan untuk mengetahui struktur bawah permukaan atau untuk tahap awal interpretasi geologi.

## 4. Input (I)

**Input** merupakan data yang digunakan dalam pekerjaan, seperti data seismik, gravitasi, magnetik, resistivitas, atau data geologi pendukung.

**Contoh:** file CSV berisi koordinat, elevasi, dan nilai gravitasi pengamatan.

## 5. Constraints (K)

**Constraints** adalah batasan yang harus dipatuhi selama proses pengolahan atau analisis.

**Contoh:** jangan mengubah data mentah, gunakan satuan tertentu, atau jangan melakukan filtering tanpa parameter yang ditentukan.

## 6. Output (O)

**Output** adalah bentuk hasil yang harus diberikan setelah tugas selesai.

**Contoh:** tabel hasil pengolahan, grafik, peta anomali, laporan Markdown, atau file CSV.

## 7. Quality Criteria (Q)

**Quality Criteria** merupakan kriteria yang digunakan untuk menentukan apakah hasil analisis sudah memenuhi standar yang ditentukan.

**Contoh:** tidak terdapat nilai koordinat yang hilang, satuan konsisten, dan hasil perhitungan dapat ditelusuri.

## 8. Optional Verification (V)

**Optional Verification** adalah langkah pemeriksaan untuk memastikan hasil pengolahan atau analisis sudah sesuai dengan data dan metode yang digunakan.

**Contoh:** membandingkan hasil perhitungan dengan perhitungan manual atau memeriksa kembali jumlah data setelah proses pengolahan.

# 5 Contoh Prompt Geofisika

## Prompt 1 — Pengolahan Data Seismik

> Bertindak sebagai **seismologist** untuk mengolah data seismik guna mengidentifikasi karakteristik sinyal dan noise pada rekaman seismik. Data ini digunakan sebagai tahap awal sebelum dilakukan analisis lebih lanjut terhadap struktur bawah permukaan. Gunakan file `seismic_data.csv` yang berisi waktu dan amplitudo sinyal. Jangan mengubah data mentah dan jangan melakukan filtering di luar parameter yang telah ditentukan. Buat grafik perbandingan sinyal sebelum dan sesudah proses pengolahan serta tabel ringkasan hasilnya. Hasil dianggap baik jika bentuk data tetap konsisten, noise dapat diidentifikasi dengan jelas, dan tidak terdapat perubahan satuan. Verifikasi hasil dengan membandingkan jumlah sampel sebelum dan sesudah pengolahan.

### Identifikasi 8 Unsur

1. **Role:** seismologist.
2. **Task:** mengolah data seismik dan mengidentifikasi noise.
3. **Context:** tahap awal sebelum analisis struktur bawah permukaan.
4. **Input:** `seismic_data.csv` berisi waktu dan amplitudo.
5. **Constraints:** tidak mengubah data mentah dan tidak melakukan filtering di luar parameter.
6. **Output:** grafik dan tabel ringkasan.
7. **Quality Criteria:** data konsisten, noise dapat diidentifikasi, dan satuan tidak berubah.
8. **Verification:** membandingkan jumlah sampel sebelum dan sesudah pengolahan.
